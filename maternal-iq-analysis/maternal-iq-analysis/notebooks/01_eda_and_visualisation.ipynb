# Introduction

This project investigates the relationship between the intelligence quotients (IQ) of children and their mothers. We are looking to observe whether maternal factors such as age and education carry any correlation in regards to a child’s intellectual ability, such that predictions of the child’s IQ can be made. The hypothesis for this project predicts a positive correlation of maternal age and education with child IQ.

Previous research has highlighted a strong association between maternal intellectual ability and the IQ of full term children (Lean, Paul, Smyser, & Rogers, 2018). Additionally, Veldkamp et al. (2020) found a strong positive relationship between maternal age and offspring’s IQ. Past research has also demonstrated children of mothers with lower educational attainment at birth were more likely to score lower on academic tests aged 10, indicating that maternal education is also a key factor in cognitive development (Lean et al., 2018).

The concept of a positive relationship between maternal and child IQ is supported by studies including Meador et al. (2011), however, this study suffers from bias, since all participating mother’s had epilepsy, limiting the applicability of results to the general population. Similarly, the studies by Lean, Paul, Smyser, & Rogers (2018), and Lean et al. (2018) were both conducted with children who were born very preterm, displaying the focus on particularly at-risk participants. Consequently, our research question approaches the mother-child relationship from a broader perspective by considering maternal age and education attainment levels, alongside child’s IQ in a more general population.

Objective: How do maternal factors (e.g., age, education level) correlate with child IQ and PPVT scores?

We are investigating the relationship between the independent variable of maternal age and education and the dependent variable of child IQ. We hypothesise that higher maternal age and better education levels correlate with higher child IQ. 

Our datasets were obtained through Kaggle.com (https://www.kaggle.com/code/jacopoferretti/mother-father-vs-child-iq-regression-analysis/notebook?scriptVersionId=212377507), a data science community with open source datasets.

This project consists of 2 datasets which were inspected for missing/null data using pandas, no data was concluded to be missing. Building on existing literature, we utilised these comprehensive datasets to explore whether the maternal factors mentioned - age and educational attainment - significantly influence a child’s IQ. The datasets were also uploaded onto Jamovi and boxplots of each variable was created in order to remove outliers. The cleaned data files were then uploaded to the notebook to be used for analysis. 

## Exploratory Data Analysis for child_iq.csv file

Child_iq.csv

The child_iq.csv dataset depicts scores from the Peabody Picture Vocabulary Test (PPVT) (Dunn & Dunn, 1981) of children at age 3 (mean = 86.9) along with maternal age and educational background. PPVT scores measure the ability to listen and understand single-word vocabulary and have been used as alternative measure of child IQ in research.

The following two blocks simply import the first dataset displaying the first 10 rows, and produce descriptive statistics. 

```python
import pandas as pd #software for data analysis and manipulation called upon multiple other times throughout the code
#read csv file for child scores at age 3
childData = pd.read_csv("../data/child_iq_outliers_removed.csv")
#display first 10 rows of childData
childData.head(10)
```

```python
#clean dataset
print(childData.isnull().sum())

# Drop the unnamed column
childData = childData.loc[:, ~childData.columns.str.contains('^Unnamed')]

# Generate descriptive statistics
descriptive_stats = childData.describe()

# Reset index to make the table clear and well-formatted
descriptive_stats = descriptive_stats.reset_index()

# Display the table
descriptive_stats
```

The '0's above the table indicate that there were no null entries in the dataset and all 400 data points were accepted. The descriptive statistics table depicts the range of the mothers' ages between 17-29 years old, with a mean of 22.7. Children's test scores show a wide distribution with a range of 40-136 and mean score 87.9. Mother's educational background was recorded with a 1-4 rating scale where 1 = no high school education, 2 = high school graduation, 3 = some college education and 4 = college graduation. Majority of mothers’ had graduated high school as indicated by a mean of 2.1.

## Visualisation of relationship between Maternal Education and Child Intelligence

### Box plot for educ_cat (mother education level) and ppvt (child test scores)

Next, the following blocks visualise the relationship between ppvt scores and mothers' education levels and child iq and mother's education individually. Box plots are used to see these relationships since maternal education level is recorded as a categorical variable in both datasets, making presentation and understanding simpler using box plots. 

```python
import pandas as pd
import matplotlib.pyplot as plt #matplot is better for visualisation here because the categorical data is integer encoded

# Load data from CSV file
data = pd.read_csv('../data/child_iq_outliers_removed.csv')

# Boxplot: Child Test Scores by Maternal Education Level using Matplotlib
plt.figure(figsize=(8, 6))

# Create a boxplot grouped by 'educ_cat (mother education level)' using Matplotlib
plt.boxplot([data[data['educ_cat (mother education level)'] == i]['ppvt (child test score)'] 
             for i in data['educ_cat (mother education level)'].unique()],
            labels=[str(i) for i in data['educ_cat (mother education level)'].unique()])

# Add title and labels
plt.title("Children's IQ Scores by Maternal Education Level")
plt.xlabel('Maternal Education Level')
plt.ylabel('Child Test Score')

# Add grid lines for clarity
plt.grid(True, which='both', axis='both', linestyle='--', color='gray', alpha=0.7)

# Show the plot
plt.tight_layout()
plt.show()
```

Superficially, the box plot shows a positive correlation between the mother's education level and child's ppvt scores, such that better educational background of the mother predicts better test scores for the child. Additionally, the variability of scores decreases with increase in maternal education level, indicating consistent scores in the data.

A box plot for the second dataset (using only 2 categories of maternal education level but still an accurate variable):

```python
import pandas as pd
import matplotlib.pyplot as plt #matplot is better for visualisation here because the categorical data is integer encoded
import numpy as np

# Load data from CSV file
data = pd.read_csv('../data/kid_iq_outliers_removed.csv')
column_data = data['kid_score (IQ score)']

# Creating a new column in dataframe storing whether or not the mother finished high school 
data['finished_high_school'] = data['mom_hs (whether or not she graduated HS'].apply(lambda x: 'Yes' if x == 1 else 'No') 
yes_iq = data[data['finished_high_school'] == 'Yes']['kid_score (IQ score)'] # filtering data to only include rows with value of 'Yes' in finished_high_school
no_iq = data[data['finished_high_school'] == 'No']['kid_score (IQ score)'] # two pandas series generated

# Calculate mean and standard deviation for "Mother Finished High School"
mean_yes = yes_iq.mean()
std_yes = yes_iq.std()
mean_no = no_iq.mean()
std_no = no_iq.std()

# Create a box plot using Matplotlib
plt.figure(figsize=(8, 6))
plt.boxplot([yes_iq, no_iq], labels=['Mother Finished High School', 'Mother Didn\'t Finish High School'])

# Add title and labels
plt.title("Children's IQ Scores by Mother's High School Completion")
plt.xlabel('High School Completion')
plt.ylabel('Children\'s IQ')

# Add grid lines
plt.grid(True, which='both', axis='both', linestyle='--', color='gray', alpha=0.7)
plt.grid(which='minor', color='lightgray', linestyle=':', linewidth=0.5, alpha=0.5)  # Minor grid lines

# Error bar with mean and standard deviation
plt.errorbar(1, mean_yes, yerr=std_yes, fmt='o', color='red', label=f'Mean ± SD (Finished HS): {mean_yes:.2f} ± {std_yes:.2f}')

# For 'No' group (Mother Didn't Finish High School)
plt.errorbar(2, mean_no, yerr=std_no, fmt='o', color='blue', label=f'Mean ± SD (Didn\'t Finish HS): {mean_no:.2f} ± {std_no:.2f}')

# Show plot
plt.legend()
plt.show() 
```

A range of IQ scores are recorded for both groups, with the box plot showing a clear increase in median IQ for mothers who have completed high school. Additionally, the calculated mean of IQ is also higher in children with mothers who have graduated high school. This is consistent with the pattern observed in the first dataset, where a better educational background is indicative of better IQ scores in children.
